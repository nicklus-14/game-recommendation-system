```python
import os
```

```python
file = open("games_metadata.json", "r", encoding="utf8")
```

```python
import json
file.seek(0)

total = 0
missing = 0

for lineno, raw in enumerate(file, start=1):
    line = raw.strip()
    if not line:
        continue
    try:
        obj = json.loads(line)
    except json.JSONDecodeError:
        continue
    total += 1
    desc = obj.get("description") if isinstance(obj, dict) else None
    if desc is None or (isinstance(desc, str) and desc.strip() == ""):
        missing += 1

print(f"Total JSON lines parsed: {total}")
print(f"Lines missing description: {missing}")
```

```python
file.seek(0)
missing_tags = 0

for lineno, raw in enumerate(file, start=1):
    line = raw.strip()
    if not line:
        continue
    try:
        obj = json.loads(line)
    except json.JSONDecodeError:
        continue
    tags = obj.get("tags") if isinstance(obj, dict) else None
    # consider missing if key absent/None, empty list/tuple, or empty string
    if tags is None or (isinstance(tags, (list, tuple)) and len(tags) == 0) or (isinstance(tags, str) and tags.strip() == ""):
        missing_tags += 1
print(f"Lines missing tags: {missing_tags}")
```
